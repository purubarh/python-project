```python
import pandas as pd
import numpy as np

def generate_sales_data(n_rows=5000):
    np.random.seed(42)  # For reproducibility

    # Generate random dates between 2023-01-01 and 2024-12-31
    start_date = pd.Timestamp("2023-01-01")
    end_date = pd.Timestamp("2024-12-31")
    dates = pd.date_range(start_date, end_date, freq='D')
    order_dates = np.random.choice(dates, size=n_rows)

    # Generate other sample columns
    product_ids = np.random.randint(1000, 1020, size=n_rows)
    quantities = np.random.randint(1, 10, size=n_rows)
    prices = np.random.uniform(10.0, 100.0, size=n_rows).round(2)

    # Create DataFrame
    df = pd.DataFrame({
        'order_date': order_dates,
        'product_id': product_ids,
        'quantity': quantities,
        'price': prices
    })

    # Optional: add derived columns
    df['year'] = df['order_date'].dt.year
    df['month'] = df['order_date'].dt.month
    df['day'] = df['order_date'].dt.day

    return df

# Example usage
df = generate_sales_data(5000)
print(df.head())
print(f"Date range: {df['order_date'].min()} to {df['order_date'].max()}")
print(df[['year', 'month', 'day']].head())
```
